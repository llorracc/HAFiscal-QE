**What it does**

1. Runs **`phase0_harness_checks()`** from `FromPandemicCode/test_asymptotic_equality_revised.py`: TM `baseline_tm_data` row count vs `economy.agents`, **`act_T`** restoration after `switch_to_counterfactual_mode`, optional per-cohort TM-init drift table (`diagnose_tm_init_stability`, harness **warmup=2**).

4. **Updates** the **Harness** row in **`history/asymptotic-equality-test-plan_revised_progress.md`** while preserving other rows (§2.2).

````python
import os
import sys
import time
from datetime import datetime, timezone
from pathlib import Path

_HERE = Path.cwd().resolve()
_CANDIDATES = [_HERE]
for _p in (_HERE,):
    for _ in range(5):
        _p = _p.parent
        _CANDIDATES.append(_p)
_seen = set()
_CANDIDATES = [c for c in _CANDIDATES if not (c in _seen or _seen.add(c))]
REPO_ROOT = None
for _cand in _CANDIDATES:
    if (_cand / "plans" / "asymptotic-equality-test-plan_revised.md").is_file():
        REPO_ROOT = _cand
        break
if REPO_ROOT is None:
    REPO_ROOT = _HERE.parent.parent
print(f"REPO_ROOT: {REPO_ROOT}")
HISTORY_DIR = REPO_ROOT / "history"
HISTORY_DIR.mkdir(parents=True, exist_ok=True)

_FPC = REPO_ROOT / "Code" / "HA-Models" / "FromPandemicCode"
if not (_FPC / "test_asymptotic_equality_revised.py").is_file():
    raise FileNotFoundError(f"Expected test_asymptotic_equality_revised.py under {_FPC}")
os.chdir(_FPC)
if str(_FPC) not in sys.path:
    sys.path.insert(0, str(_FPC))
_HM = _FPC.parent
if str(_HM) not in sys.path:
    sys.path.insert(0, str(_HM))

PLAN_STEM = "asymptotic-equality-test-plan_revised"
STEP_NAME_FILE = "Harness"
PROGRESS_NAME = "asymptotic-equality-test-plan_revised_progress.md"

print("REPO_ROOT:", REPO_ROOT)
print("FromPandemicCode:", _FPC)
````

````python
# Parametrization: Reduced_Run (plan default) or Smoke_Test for a fast wiring check
HARNESS_PARAMS = {
    "parametrization": "Reduced_Run",  # or "Smoke_Test"
}

HARNESS_PARAMS
````

````python
import importlib

import test_asymptotic_equality_revised as harness_mod

importlib.reload(harness_mod)
harness_mod.RUN_PARAMETRIZATION = str(HARNESS_PARAMS["parametrization"])

wall_start = datetime.now(timezone.utc)
t0 = time.perf_counter()

phase_out = harness_mod.phase0_harness_checks()

wall_end = datetime.now(timezone.utc)
elapsed_s = time.perf_counter() - t0

phase_out
````

````python
import pandas as pd

summary = phase_out.get("per_type_init_stability") or {}
rows = summary.get("per_type") or []

if rows:
    df = pd.DataFrame(rows)
    # Flatten for display: key columns first
    cols = [
        "type_index",
        "AgentCount",
        "init_looks_stable",
        "max_m_drift_early",
        "max_p_drift_early",
        "max_u_drift_pp_early",
        "max_varlog_m_drift_early",
        "max_varlog_p_drift_early",
    ]
    cols = [c for c in cols if c in df.columns]
    display(df[cols])
else:
    print("(No per-type init stability table: single-type economy.)")

wm = summary.get("weighted_mean") or {}
if wm:
    display(pd.DataFrame([wm]))

print("\n=== Harness summary ===")
print("n_agents:", phase_out.get("n_agents"))
print("act_T:", phase_out.get("act_T"))
print("act_T after switch (before fix):", phase_out.get("act_T_after_switch_before_fix"))
if rows:
    n_ok = sum(1 for r in rows if r.get("init_looks_stable"))
    print(f"per-cohort init_looks_stable: {n_ok}/{len(rows)}")
````

````python
def _fmt_duration(seconds: float) -> str:
    s = int(round(seconds))
    m, s = divmod(s, 60)
    h, m = divmod(m, 60)
    if h:
        return f"{h}h {m}m {s}s"
    if m:
        return f"{m}m {s}s"
    return f"{s}s"


def _report_basename_utc() -> str:
    base = wall_start.strftime("%Y%m%dT%H%M")
    name = f"{PLAN_STEM}_{STEP_NAME_FILE}_{base}.md"
    p = HISTORY_DIR / name
    n = 2
    while p.is_file():
        name = f"{PLAN_STEM}_{STEP_NAME_FILE}_{base}_{n}.md"
        p = HISTORY_DIR / name
        n += 1
    return name


rows_pt = (phase_out.get("per_type_init_stability") or {}).get("per_type") or []
init_all_ok = (
    all(r.get("init_looks_stable") for r in rows_pt) if rows_pt else True
)
harness_ok = True  # phase0 uses assertions; we got here without exception

report_bn = _report_basename_utc()
report_path = HISTORY_DIR / report_bn

lines = []
lines.append(f"# {PLAN_STEM} — {STEP_NAME_FILE}")
lines.append("")
lines.append(f"**Generated (UTC):** {wall_end.isoformat()}")
lines.append("")
lines.append("**Plan:** `plans/20260404-1746h_asymptotic-equality-test-plan_revised.md` (Harness §2)")
lines.append("")
lines.append("## Parametrization")
lines.append("")
lines.append(f"- **Parametrization:** `{HARNESS_PARAMS['parametrization']}`")
lines.append(
    "- **CLI equivalent:** "
    f"`python test_asymptotic_equality_revised.py --phase 0 "
    f"--parametrization {HARNESS_PARAMS['parametrization']}`"
)
lines.append("")
lines.append("## Results")
lines.append("")
lines.append(f"- **n_types (agents):** {phase_out.get('n_agents')}")
lines.append(f"- **act_T (restored):** {phase_out.get('act_T')}")
lines.append(
    f"- **act_T immediately after switch (before fix):** "
    f"{phase_out.get('act_T_after_switch_before_fix')}"
)
lines.append(f"- **Harness assertions:** {'PASS' if harness_ok else 'FAIL'}")
lines.append(
    f"- **All cohorts init_looks_stable:** {'PASS' if init_all_ok else 'WARN (see notebook table)'}"
)
lines.append("")
wm = (phase_out.get("per_type_init_stability") or {}).get("weighted_mean") or {}
if wm:
    lines.append("### AgentCount-weighted mean drift (early periods)")
    lines.append("")
    for k, v in wm.items():
        lines.append(f"- **{k}:** {v}")
    lines.append("")
if rows_pt:
    lines.append("### Per cohort")
    lines.append("")
    lines.append("| i | N | init_ok | max|dm|/m0 | max|dp|/p0 | max|du|pp |")
    lines.append("|---|------|---------|-----------|-----------|----------|")
    for r in rows_pt:
        lines.append(
            f"| {r.get('type_index')} | {r.get('AgentCount')} | "
            f"{r.get('init_looks_stable')} | {r.get('max_m_drift_early')} | "
            f"{r.get('max_p_drift_early')} | {r.get('max_u_drift_pp_early')} |"
        )
    lines.append("")
lines.append("## Interpretation")
lines.append("")
lines.append(
    "Harness verifies per-type TM indexing and `act_T` after counterfactual switch. "
    "Per-cohort `init_looks_stable` can be **N** for small `AgentCount` (noisy unemployment share); "
    "see `verify_four_methods_agreement.diagnose_tm_init_stability` thresholds."
)
lines.append("")
lines.append("## Step timing")
lines.append("")
lines.append(f"- **Step:** Harness")
lines.append(f"- **Wall-clock start (UTC):** {wall_start.strftime('%Y-%m-%dT%H:%M:%SZ')}")
lines.append(f"- **Wall-clock end (UTC):** {wall_end.strftime('%Y-%m-%dT%H:%M:%SZ')}")
lines.append(f"- **Duration:** {_fmt_duration(elapsed_s)} ({elapsed_s:.1f} s)")
lines.append("")

report_path.write_text("\n".join(lines), encoding="utf-8")
print("Wrote:", report_path.relative_to(REPO_ROOT))
rel_report = report_path.relative_to(REPO_ROOT).as_posix()
````

````python
progress_path = HISTORY_DIR / PROGRESS_NAME
now_iso = wall_end.strftime("%Y-%m-%dT%H:%M:%SZ")
start_iso = wall_start.strftime("%Y-%m-%dT%H:%M:%SZ")
dur_human = _fmt_duration(elapsed_s)
overall_note = "done" if (harness_ok and init_all_ok) else "done (see cohort warnings)"

if progress_path.is_file():
    plines = progress_path.read_text(encoding="utf-8").splitlines()
    out_lines = []
    for line in plines:
        if line.startswith("**Last update (UTC):**"):
            out_lines.append(f"**Last update (UTC):** {now_iso}")
        elif line.startswith("**Overall run status:**"):
            out_lines.append(
                "**Overall run status:** in progress (Harness notebook; Gatekeeper unchanged)"
            )
        elif line.startswith("| Harness |"):
            out_lines.append(
                f"| Harness | {overall_note} | {start_iso} | {now_iso} | "
                f"{dur_human} ({elapsed_s:.1f} s) | {rel_report} |"
            )
        else:
            out_lines.append(line)
    # Extend footer if not already mentioning Harness notebook
    text_out = "\n".join(out_lines)
    if "Harness_Asymptotic_Equality.ipynb" not in text_out:
        text_out = text_out.rstrip() + (
            "\n\n*Harness row updated by `Code/HA-Models/Harness_Asymptotic_Equality.ipynb`.*\n"
        )
    progress_path.write_text(text_out + "\n", encoding="utf-8")
else:
    tbl = [
        "| Step | Status | Started (UTC) | Ended (UTC) | Duration | Report |",
        "|------|--------|---------------|-------------|----------|--------|",
        "| Gatekeeper | pending |  |  |  |  |",
        f"| Harness | {overall_note} | {start_iso} | {now_iso} | "
        f"{dur_human} ({elapsed_s:.1f} s) | {rel_report} |",
        "| Multi-type baseline | pending |  |  |  |  |",
        "| No-recession policies | pending |  |  |  |  |",
        "| Recession suite | pending |  |  |  |  |",
        "| AD feedback loop | pending |  |  |  |  |",
        "| Convergence sweep | pending |  |  |  |  |",
    ]
    nl = chr(10)
    body = (
        f"# {PLAN_STEM} — progress{nl}{nl}"
        f"**Last update (UTC):** {now_iso}{nl}"
        f"**Overall run status:** in progress{nl}"
        f"**Currently running:** none{nl}{nl}"
        f"## Steps{nl}{nl}"
        f"{nl.join(tbl)}{nl}{nl}"
        f"---{nl}"
        "*Harness row created by `Code/HA-Models/Harness_Asymptotic_Equality.ipynb`.*"
        f"{nl}"
    )
    progress_path.write_text(body, encoding="utf-8")

print("Wrote:", progress_path.relative_to(REPO_ROOT))
````
